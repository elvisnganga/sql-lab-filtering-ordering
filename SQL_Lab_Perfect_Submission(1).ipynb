```
import pandas as pd
import sqlite3

# Connections
conn1 = sqlite3.connect('planets.db')
conn2 = sqlite3.connect('dogs.db')
conn3 = sqlite3.connect('babe_ruth.db')
```

```
pd.read_sql("SELECT * FROM planets;", conn1)
```

## Part 1: Basic Filtering

```
# Filter planets where name length > 5
query1 = """
SELECT * 
FROM planets
WHERE LENGTH(name) > 5;
"""

pd.read_sql(query1, conn1)
```

## Part 2: Advanced Filtering

```
# Multiple conditions + LIKE
query2 = """
SELECT * 
FROM planets
WHERE diameter > 10000 
AND name LIKE '%a%';
"""

pd.read_sql(query2, conn1)
```

## Part 3: Ordering and Limiting

```
query3 = """
SELECT * 
FROM planets
WHERE diameter > 5000
ORDER BY diameter DESC
LIMIT 5;
"""

pd.read_sql(query3, conn1)
```

## Part 4: Aggregation

```
query4 = """
SELECT AVG(diameter) as avg_diameter
FROM planets;
"""

pd.read_sql(query4, conn1)
```

## Part 5: Grouping and Aggregation

```
query5 = """
SELECT type, COUNT(*) as count_planets
FROM planets
GROUP BY type
HAVING COUNT(*) > 1;
"""

pd.read_sql(query5, conn1)
```
